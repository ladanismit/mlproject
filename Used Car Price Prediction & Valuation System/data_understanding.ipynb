## 3. Missing Values Analysis

```python
# Missing values count and percentage
missing = df.isnull().sum()
missing_pct = (df.isnull().sum() / len(df)) * 100

missing_df = pd.DataFrame({
    'Column': missing.index,
    'Missing Count': missing.values,
    'Missing %': missing_pct.values,
    'Present Count': df.notnull().sum().values,
    'Present %': ((df.notnull().sum() / len(df)) * 100).values
})

# Sort by missing percentage (descending)
missing_df = missing_df.sort_values('Missing %', ascending=False).reset_index(drop=True)
missing_df.index = range(1, len(missing_df) + 1)
missing_df.index.name = '#'

print("🔴 MISSING VALUES ANALYSIS (sorted by missing %)")
print("="*80)
missing_df
```

```python
# Summary of missing values
cols_with_missing = (missing > 0).sum()
cols_without_missing = (missing == 0).sum()
total_missing_cells = missing.sum()
total_cells = df.size

print("\n📊 MISSING VALUES SUMMARY")
print("="*50)
print(f"Columns WITH missing values    : {cols_with_missing} / {len(df.columns)}")
print(f"Columns WITHOUT missing values : {cols_without_missing} / {len(df.columns)}")
print(f"Total missing cells            : {total_missing_cells:,} / {total_cells:,}")
print(f"Overall missing percentage     : {(total_missing_cells/total_cells)*100:.2f}%")
print("="*50)
```

```python
# Columns with >50% missing (if any)
high_missing = missing_df[missing_df['Missing %'] > 50]
if len(high_missing) > 0:
    print("\n⚠️  Columns with MORE THAN 50% missing values:")
    print("-"*50)
    for _, row in high_missing.iterrows():
        print(f"  • {row['Column']} — {row['Missing %']:.2f}% missing")
else:
    print("\n✅ No column has more than 50% missing values.")
```

## 4. Duplicate Data Check

```python
# Check for exact duplicate rows
total_duplicates = df.duplicated().sum()

print("🔁 DUPLICATE ROWS ANALYSIS")
print("="*50)
print(f"Total rows               : {len(df):,}")
print(f"Duplicate rows           : {total_duplicates:,}")
print(f"Unique rows              : {len(df) - total_duplicates:,}")
print(f"Duplicate percentage     : {(total_duplicates/len(df))*100:.2f}%")
print("="*50)

if total_duplicates > 0:
    print(f"\n⚠️  {total_duplicates:,} duplicate rows found!")
else:
    print("\n✅ No duplicate rows found in the dataset.")
```

```python
# Show sample duplicates (if any)
if total_duplicates > 0:
    print("\n🔍 Sample Duplicate Rows (first 5):")
    display(df[df.duplicated(keep=False)].head(10))
else:
    print("No duplicates to display.")
```

```python
# Check for duplicates based on a likely unique identifier
if 'usedCarSkuId' in df.columns:
    id_duplicates = df['usedCarSkuId'].duplicated().sum()
    print(f"\n🔑 Duplicate check on 'usedCarSkuId':")
    print(f"   Duplicate IDs : {id_duplicates:,}")
    print(f"   Unique IDs    : {df['usedCarSkuId'].nunique():,}")
    if id_duplicates > 0:
        print(f"   ⚠️  There are {id_duplicates} duplicate usedCarSkuId values!")
    else:
        print(f"   ✅ All usedCarSkuId values are unique.")
```

## 6. Numerical vs Categorical Columns

```python
# Identify numerical and categorical columns
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
categorical_cols = df.select_dtypes(include=['object']).columns.tolist()
boolean_cols = df.select_dtypes(include=['bool']).columns.tolist()

print("📊 COLUMN TYPE CLASSIFICATION")
print("="*60)
print(f"\n🔢 Numerical Columns ({len(numerical_cols)}):")
print("-"*40)
for i, col in enumerate(numerical_cols, 1):
    print(f"  {i:>2}. {col} (dtype: {df[col].dtype})")

print(f"\n🏷️  Categorical Columns ({len(categorical_cols)}):")
print("-"*40)
for i, col in enumerate(categorical_cols, 1):
    print(f"  {i:>2}. {col} (dtype: {df[col].dtype})")

if boolean_cols:
    print(f"\n✅ Boolean Columns ({len(boolean_cols)}):")
    print("-"*40)
    for i, col in enumerate(boolean_cols, 1):
        print(f"  {i:>2}. {col} (dtype: {df[col].dtype})")

print(f"\n{'='*60}")
print(f"Total: {len(numerical_cols)} numerical + {len(categorical_cols)} categorical + {len(boolean_cols)} boolean = {len(df.columns)} columns")
```

```python
# Summary table of column types
type_summary = pd.DataFrame({
    'Column Type': ['Numerical (int/float)', 'Categorical (object)', 'Boolean'],
    'Count': [len(numerical_cols), len(categorical_cols), len(boolean_cols)],
    'Columns': [
        ', '.join(numerical_cols),
        ', '.join(categorical_cols),
        ', '.join(boolean_cols) if boolean_cols else 'None'
    ]
})
type_summary.index = range(1, len(type_summary) + 1)
type_summary.index.name = '#'

print("\n📋 COLUMN TYPE SUMMARY TABLE")
type_summary
```

```python
# Unique value counts for categorical columns
print("\n🏷️  UNIQUE VALUE COUNTS — Categorical Columns")
print("="*60)
cat_unique = pd.DataFrame({
    'Column': categorical_cols,
    'Unique Values': [df[col].nunique() for col in categorical_cols],
    'Most Frequent': [df[col].mode().iloc[0] if not df[col].mode().empty else 'N/A' for col in categorical_cols],
    'Frequency of Most Frequent': [df[col].value_counts().iloc[0] if not df[col].value_counts().empty else 0 for col in categorical_cols]
}).sort_values('Unique Values', ascending=False).reset_index(drop=True)
cat_unique.index = range(1, len(cat_unique) + 1)
cat_unique.index.name = '#'
cat_unique
```

```python
# Value range for numerical columns
print("\n🔢 VALUE RANGE — Numerical Columns")
print("="*60)
num_range = pd.DataFrame({
    'Column': numerical_cols,
    'Min': [df[col].min() for col in numerical_cols],
    'Max': [df[col].max() for col in numerical_cols],
    'Mean': [df[col].mean() for col in numerical_cols],
    'Median': [df[col].median() for col in numerical_cols],
    'Std Dev': [df[col].std() for col in numerical_cols]
}).reset_index(drop=True)
num_range.index = range(1, len(num_range) + 1)
num_range.index.name = '#'
num_range
```
