```python
import numpy as np
import tensorflow as tf
from sklearn.metrics import classification_report, confusion_matrix

from src.config import BEST_MODEL_PATH, CLASSES
from src.dataset import get_datasets

print("="*70)
print("DOCVISION MODEL VALIDATION")
print("="*70)

# Load datasets
train_ds, val_ds, test_ds = get_datasets()

# Load saved model
model = tf.keras.models.load_model(str(BEST_MODEL_PATH))

print("\nModel loaded successfully!")
print(model.summary())

print("\nEvaluating on Test Dataset...")

loss, accuracy = model.evaluate(test_ds, verbose=1)

print("\n" + "="*70)
print(f"Test Loss     : {loss:.4f}")
print(f"Test Accuracy : {accuracy*100:.2f}%")
print("="*70)

# Classification report
y_true = []
y_pred = []

for images, labels in test_ds:
    predictions = model.predict(images, verbose=0)
    predictions = np.argmax(predictions, axis=1)

    y_true.extend(labels.numpy())
    y_pred.extend(predictions)

print("\nClassification Report\n")
print(classification_report(y_true, y_pred, target_names=CLASSES))

print("\nConfusion Matrix\n")
print(confusion_matrix(y_true, y_pred))
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

from src.preprocessing import preprocess_image
from src.config import CLASSES

image_path = "/kaggle/input/datasets/smitladani/docvision-dataset/docvision-dataset/Invoice/X51005442344.jpg"

image = preprocess_image(image_path)

prediction = model.predict(np.expand_dims(image, axis=0), verbose=0)[0]

predicted_class = CLASSES[np.argmax(prediction)]
confidence = np.max(prediction)

print("="*70)
print("Single Image Prediction")
print("="*70)
print("Predicted Class :", predicted_class)
print("Confidence      :", confidence*100)

print("\nProbabilities")

for cls, prob in zip(CLASSES, prediction):
    print(f"{cls:<10}: {prob*100:.2f}%")

img = cv2.imread(image_path)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8,8))
plt.imshow(img)
plt.title(f"{predicted_class} ({confidence*100:.2f}%)")
plt.axis("off")
plt.show()
```
