### 4. Missing Values
Are there any missing values in the data?

```python
train_missing = train.isnull().sum()
train_missing = train_missing[train_missing > 0]

if train_missing.empty:
    print('Training Data: No missing values found!')
else:
    print('--- Training Data: Missing Values ---')
    missing_df = pd.DataFrame({
        'Missing Count': train_missing,
        'Missing %': round((train_missing / len(train)) * 100, 2)
    }).sort_values('Missing Count', ascending=False)
    print(missing_df)
    print(f'\nTotal columns with missing values: {len(train_missing)} out of {train.shape[1]}')
```

```python
test_missing = test.isnull().sum()
test_missing = test_missing[test_missing > 0]

if test_missing.empty:
    print('Test Data: No missing values found!')
else:
    print('--- Test Data: Missing Values ---')
    missing_df = pd.DataFrame({
        'Missing Count': test_missing,
        'Missing %': round((test_missing / len(test)) * 100, 2)
    }).sort_values('Missing Count', ascending=False)
    print(missing_df)
    print(f'\nTotal columns with missing values: {len(test_missing)} out of {test.shape[1]}')
```

### 5. Duplicate Entries
Are there any duplicate rows in the data?

```python
train_dupes = train.duplicated().sum()
test_dupes = test.duplicated().sum()

print(f'Training Data: {train_dupes} duplicate rows')
print(f'Test Data    : {test_dupes} duplicate rows')
```

### 6. Key Observations

| Observation | Detail |
|---|---|
| **New numerical features** | `EXT_SOURCE_1`, `EXT_SOURCE_2`, `EXT_SOURCE_3` (normalized external credit scores), `DAYS_BIRTH` (negative days from application), `DAYS_EMPLOYED` (negative days from application, with anomalies) |
| **New categorical features** | `ORGANIZATION_TYPE` (58 unique values), `NAME_HOUSING_TYPE` (6 unique values) |
| **High missing rate** | `EXT_SOURCE_1` has ~56% missing values |
| **Anomaly** | `DAYS_EMPLOYED` contains a suspicious value of 365243 (positive) for many rows — likely represents unemployed/pensioner |

```python
# Check the DAYS_EMPLOYED anomaly
anomaly_count = (train['DAYS_EMPLOYED'] == 365243).sum()
print(f'DAYS_EMPLOYED == 365243 (anomaly): {anomaly_count} rows ({anomaly_count/len(train)*100:.2f}%)')
print(f'\nDAYS_EMPLOYED stats (all):')
print(train['DAYS_EMPLOYED'].describe())
print(f'\nDAYS_EMPLOYED stats (excluding anomaly):')
print(train.loc[train['DAYS_EMPLOYED'] != 365243, 'DAYS_EMPLOYED'].describe())
```
